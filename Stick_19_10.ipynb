```python

import os
import requests

# Identify the EGOFALLS dataset using its persistent DOI
api_url = "https://dataverse.nl/api/datasets/:persistentId/"
params = {"persistentId": "doi:10.34894/HO5GE3"}

# Retrieve the official file listing, including file IDs and expected sizes
response = requests.get(api_url, params=params, timeout=60)
response.raise_for_status()
dataset = response.json()
dataset_files = dataset["data"]["latestVersion"]["files"]

# Select the four additional archives needed for fall-detection training
required_files = ["S_FI.zip", "S_L.zip", "S_Q.zip", "S_XL.zip"]

# Set the Google Drive folder where raw datasets will be stored
save_folder = (
    "/content/drive/MyDrive/fall_stress_project/raw_temp"
)
os.makedirs(save_folder, exist_ok=True)

# Create a lookup table using the official Dataverse file listing
file_lookup = {
    item.get("label", item["dataFile"].get("filename")): item["dataFile"]
    for item in dataset_files
}

# Download each archive sequentially to avoid simultaneous large downloads
for filename in required_files:
    if filename not in file_lookup:
        print(f"File not found in Dataverse listing: {filename}")
        continue

    # Get the official file ID and expected file size
    file_info = file_lookup[filename]
    file_id = file_info["id"]
    expected_size = file_info["filesize"]

    # Define the final path and a temporary path for incomplete downloads
    save_path = os.path.join(save_folder, filename)
    partial_path = save_path + ".part"

    # Skip downloading if a complete, correctly sized file already exists
    if os.path.exists(save_path):
        if os.path.getsize(save_path) == expected_size:
            print(f"Already downloaded and verified: {filename}")
            continue

    # Download from the official Dataverse file endpoint
    download_url = (
        f"https://dataverse.nl/api/access/datafile/{file_id}"
    )

    print(f"\nDownloading {filename} (ID: {file_id})...")

    with requests.get(
        download_url,
        stream=True,
        timeout=(30, 300)
    ) as download_response:
        download_response.raise_for_status()

        # Write incoming chunks to a temporary file on Google Drive
        with open(partial_path, "wb") as output_file:
            for chunk in download_response.iter_content(
                chunk_size=8 * 1024 * 1024
            ):
                if chunk:
                    output_file.write(chunk)

    # Verify that the downloaded file matches Dataverse's reported size
    downloaded_size = os.path.getsize(partial_path)

    if downloaded_size != expected_size:
        raise RuntimeError(
            f"Size mismatch for {filename}. "
            f"Expected {expected_size} bytes, "
            f"received {downloaded_size} bytes. "
            "The incomplete file was kept as a .part file."
        )

    # Rename the verified temporary file to its final ZIP filename
    os.replace(partial_path, save_path)

    print(f"Completed: {filename}")
    print(f"Size: {downloaded_size / (1024**3):.2f} GB")
    print(f"Saved at: {save_path}")

print("\nAll four additional archive downloads are complete.")
```

```python

import os
import zipfile

# Define the Google Drive folder containing the downloaded ZIP archives
raw_folder = (
    "/content/drive/MyDrive/fall_stress_project/raw_temp"
)

# Define temporary local storage for extracted datasets
extract_base = "/content/egofalls_extracted"

# Select only archives that we have verified and can extract independently
selected_archives = [
    "S_HB.zip",
    "S_FI.zip",
    "S_L.zip",
    "S_F.zip"
]

# Create the local extraction directory if it does not exist
os.makedirs(extract_base, exist_ok=True)

# Extract each selected archive into its own folder
for filename in selected_archives:
    zip_path = os.path.join(raw_folder, filename)
    dataset_name = os.path.splitext(filename)[0]
    output_folder = os.path.join(extract_base, dataset_name)

    # Ensure the source ZIP exists before starting extraction
    if not os.path.exists(zip_path):
        print(f"Skipping missing archive: {filename}")
        continue

    # Skip extraction if the destination folder already exists and is non-empty
    if os.path.isdir(output_folder) and os.listdir(output_folder):
        print(f"Already extracted: {dataset_name}")
        continue

    # Open the archive and extract its contents into the destination folder
    with zipfile.ZipFile(zip_path, "r") as archive:
        # Check for potentially unsafe paths before extracting files
        for member in archive.infolist():
            target_path = os.path.realpath(
                os.path.join(output_folder, member.filename)
            )
            if not (
                target_path == os.path.realpath(output_folder)
                or target_path.startswith(
                    os.path.realpath(output_folder) + os.sep
                )
            ):
                raise ValueError(
                    f"Unsafe archive path detected: {member.filename}"
                )

        # Extract all archive contents
        archive.extractall(output_folder)

    print(f"Extraction completed: {dataset_name}")

# Display the folders created in temporary storage
print("\nExtracted dataset folders:")
print(os.listdir(extract_base))
```

```python

import os
import cv2

# Define the extracted EGOFALLS dataset directory
base = "/content/egofalls_extracted"

# Define the output directory for sampled frames
output_base = "/content/egofalls_frames"

# Set the frame sampling rate to 2 frames per second
sample_fps = 2

# Define supported video file extensions
video_extensions = (".mp4", ".mov", ".avi", ".mkv")

# Select one sample video for the initial preprocessing test
sample_video = (
    "/content/egofalls_extracted/S_L/S_L/Indoor/"
    "nonfalls/Lying/lying-bed/Waist/S_L_lying-bed_Waist_5.mp4"
)

# Create an output folder for frames from the sample video
video_name = os.path.splitext(os.path.basename(sample_video))[0]
frame_output = os.path.join(output_base, video_name)
os.makedirs(frame_output, exist_ok=True)

# Open the video and determine its frame rate
cap = cv2.VideoCapture(sample_video)
fps = cap.get(cv2.CAP_PROP_FPS)

if not cap.isOpened() or fps <= 0:
    raise RuntimeError("Could not open the video or read its FPS.")

# Calculate how many original frames to skip between samples
frame_interval = max(1, round(fps / sample_fps))
frame_index = 0
saved_count = 0

# Read the video and save one frame at each sampling interval
while True:
    success, frame = cap.read()

    if not success:
        break

    if frame_index % frame_interval == 0:
        # Save the sampled frame as a JPEG image
        frame_path = os.path.join(
            frame_output,
            f"frame_{saved_count:04d}.jpg"
        )
        cv2.imwrite(frame_path, frame)
        saved_count += 1

    frame_index += 1

# Release the video after processing
cap.release()

# Display the output location and number of saved frames
print("Sample video:", video_name)
print("Frames saved:", saved_count)
print("Output folder:", frame_output)
```

```python

import os
import pandas as pd

# Extracted EGOFALLS datasets ka main folder
base = "/content/egofalls_extracted"

# Metadata CSV ko Google Drive mein save karne ka path
output_path = (
    "/content/drive/MyDrive/fall_stress_project/"
    "processed/camera2_egofalls/video_metadata.csv"
)

# Video formats jinhe hum process karenge
video_extensions = (".mp4", ".mov", ".avi", ".mkv")

# Har video ki information store karne ke liye list
records = []

# Extracted dataset ke andar saare video files search karo
for root, dirs, files in os.walk(base):
    for filename in files:
        if not filename.lower().endswith(video_extensions):
            continue

        # Video ka complete path aur relative path nikalo
        video_path = os.path.join(root, filename)
        relative_path = os.path.relpath(video_path, base)

        # Folder names se fall/non-fall label identify karo
        parts = relative_path.split(os.sep)
        lower_parts = [part.lower() for part in parts]

        if "falls" in lower_parts:
            label = "fall"
        elif "nonfalls" in lower_parts:
            label = "non-fall"
        else:
            label = "unknown"

        # Activity aur camera position folder names se identify karo
        category_index = next(
            (i for i, part in enumerate(lower_parts)
             if part in ("falls", "nonfalls")),
            None
        )

        activity = (
            parts[category_index + 1]
            if category_index is not None
            and category_index + 1 < len(parts)
            else "unknown"
        )

        camera_position = next(
            (part for part in lower_parts
             if part in ("neck", "waist")),
            "unknown"
        )

        # Video ki metadata ek record mein store karo
        records.append({
            "video_name": filename,
            "video_path": video_path,
            "dataset": parts[0],
            "label": label,
            "activity": activity,
            "camera_position": camera_position
        })

# Saare records ko table mein convert karo
df = pd.DataFrame(records)

# Output folder create karke metadata CSV save karo
os.makedirs(os.path.dirname(output_path), exist_ok=True)
df.to_csv(output_path, index=False)

# Total videos aur label counts display karo
print("Total videos:", len(df))
print("\nLabel counts:")
print(df["label"].value_counts())

print("\nMetadata saved at:", output_path)
print("\nFirst 5 records:")
print(df.head())
```

```python
import os
import cv2
import numpy as np
import pandas as pd

# Load the processed metadata for the balanced dataset
csv_path = (
    "/content/drive/MyDrive/fall_stress_project/"
    "processed/camera2_egofalls/metadata_balanced_100_processed.csv"
)
results_df = pd.read_csv(csv_path)

# Define sequence and image dimensions
sequence_length = 16
image_size = (224, 224)

# Store sequences and their associated metadata
all_sequences = []
sequence_labels = []
source_video_ids = []
sequence_info = []

# Process each video independently
for video_idx, row in results_df.iterrows():
    frame_folder = row["frame_folder"]
    label = row["label"]
    video_name = row["video_name"]

    # List frames in chronological order
    frame_files = sorted([
        f for f in os.listdir(frame_folder)
        if f.lower().endswith(".jpg")
    ])

    frames = []

    # Load and standardize each frame
    for filename in frame_files:
        frame_path = os.path.join(frame_folder, filename)
        image = cv2.imread(frame_path)

        if image is not None:
            image = cv2.resize(image, image_size)
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
            frames.append(image)

    # Skip videos that have no readable frames
    if len(frames) < sequence_length:
        continue

    # Convert frames to a NumPy array
    frames = np.asarray(frames, dtype=np.uint8)

    # Create non-overlapping sequences of 16 frames
    num_sequences = len(frames) // sequence_length

    for seq_idx in range(num_sequences):
        start = seq_idx * sequence_length
        end = start + sequence_length
        sequence = frames[start:end]

        # Store the sequence, its label, and its source video
        all_sequences.append(sequence)
        sequence_labels.append(1 if label == "fall" else 0)
        source_video_ids.append(video_name)

        sequence_info.append({
            "video_name": video_name,
            "sequence_index": seq_idx,
            "label": label,
            "frame_count": sequence_length
        })

    # Report progress after every 10 videos
    if (video_idx + 1) % 10 == 0:
        print(f"Processed {video_idx + 1}/100 videos")

# Convert generated data into NumPy arrays
X = np.asarray(all_sequences, dtype=np.uint8)
y = np.asarray(sequence_labels, dtype=np.int64)
video_ids = np.asarray(source_video_ids)
sequence_metadata = pd.DataFrame(sequence_info)

# Display the generated dataset summary
print("\n--- SEQUENCE DATASET SUMMARY ---")
print("Total sequences:", len(X))
print("Sequence array shape:", X.shape)
print("Labels array shape:", y.shape)
print("Unique source videos:", len(np.unique(video_ids)))
print("Fall sequences:", int(np.sum(y == 1)))
print("Non-fall sequences:", int(np.sum(y == 0)))
print("Videos skipped for having fewer than 16 readable frames:",
      len(results_df) - len(results_df[
          results_df["frames_saved"] >= sequence_length
      ]))

# Save sequence metadata to Google Drive
metadata_output = (
    "/content/drive/MyDrive/fall_stress_project/"
    "processed/camera2_egofalls/sequence_metadata_100.csv"
)
sequence_metadata.to_csv(metadata_output, index=False)

print("Sequence metadata saved at:", metadata_output)
```

```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2

# Load MobileNetV2 pretrained on ImageNet without its classification head
cnn = MobileNetV2(
    weights="imagenet",
    include_top=False,
    pooling="avg",
    input_shape=(224, 224, 3)
)

# Freeze CNN weights for the initial experiment
cnn.trainable = False

# Display the model configuration
print("CNN model loaded:", cnn.name)
print("Input shape:", cnn.input_shape)
print("Output shape:", cnn.output_shape)
print("Trainable parameters:", sum(
    tf.keras.backend.count_params(w)
    for w in cnn.trainable_weights
))
print("Non-trainable parameters:", sum(
    tf.keras.backend.count_params(w)
    for w in cnn.non_trainable_weights
))

# Confirm GPU availability
print("Available GPUs:", tf.config.list_physical_devices("GPU"))
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model

# Define the input as a sequence of 16 RGB frames
sequence_input = layers.Input(
    shape=(16, 224, 224, 3),
    name="video_sequence"
)

# Apply the frozen CNN independently to each frame
frame_features = layers.TimeDistributed(
    cnn,
    name="frame_feature_extractor"
)(sequence_input)

# Learn temporal patterns across the 16 frame features
temporal_features = layers.GRU(
    128,
    name="temporal_gru"
)(frame_features)

# Add regularization to reduce overfitting
x = layers.Dropout(0.3, name="dropout")(temporal_features)

# Predict the probability that the sequence contains a fall
output = layers.Dense(
    1,
    activation="sigmoid",
    name="fall_probability"
)(x)

# Build the complete sequence classification model
model = Model(
    inputs=sequence_input,
    outputs=output,
    name="FallDetection_CNN_GRU"
)

# Configure the model for binary classification
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-3),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall")
    ]
)

# Display the model architecture and parameter counts
model.summary()
```

```python
import os
import numpy as np
import pandas as pd
import tensorflow as tf

# Load the saved sequence metadata and split assignments
data_dir = (
    "/content/drive/MyDrive/fall_stress_project/"
    "processed/camera2_egofalls/sequences_100_test"
)

split_path = os.path.join(data_dir, "sequence_metadata_split.csv")
split_df = pd.read_csv(split_path)

# Load sequences as a memory-mapped array
X = np.load(
    os.path.join(data_dir, "X_sequences.npy"),
    mmap_mode="r"
)
y = np.load(os.path.join(data_dir, "y_labels.npy"))

# Create indices for each video-wise split
train_idx = np.flatnonzero(split_df["split"].eq("train").to_numpy())
val_idx = np.flatnonzero(split_df["split"].eq("validation").to_numpy())
test_idx = np.flatnonzero(split_df["split"].eq("test").to_numpy())

# Build a batched TensorFlow dataset that normalizes images on demand
def make_dataset(indices, batch_size=2, shuffle=False):
    # Create a dataset of sequence indices and their labels
    dataset = tf.data.Dataset.from_tensor_slices((indices, y[indices]))

    # Shuffle only the training data
    if shuffle:
        dataset = dataset.shuffle(
            buffer_size=len(indices),
            seed=42,
            reshuffle_each_iteration=True
        )

    # Load and normalize each selected sequence
    def load_sequence(index, label):
        sequence = tf.numpy_function(
            lambda i: np.asarray(X[int(i)], dtype=np.float32) / 255.0,
            [index],
            Tout=tf.float32
        )
        sequence.set_shape((16, 224, 224, 3))
        label = tf.cast(label, tf.float32)
        return sequence, label

    dataset = dataset.map(
        load_sequence,
        num_parallel_calls=1
    )

    # Group sequences into small batches
    return dataset.batch(batch_size).prefetch(1)

# Create training, validation, and test datasets
train_ds = make_dataset(train_idx, batch_size=2, shuffle=True)
val_ds = make_dataset(val_idx, batch_size=2)
test_ds = make_dataset(test_idx, batch_size=2)

# Inspect one batch without loading the entire dataset
for batch_x, batch_y in train_ds.take(1):
    print("Batch image shape:", batch_x.shape)
    print("Batch label shape:", batch_y.shape)
    print("Pixel range:", float(tf.reduce_min(batch_x)),
          "to", float(tf.reduce_max(batch_x)))
    print("Labels in batch:", batch_y.numpy())
```

```python
import os
import cv2
import shutil
import pandas as pd
from tqdm.auto import tqdm

# Define input and output paths
split_path = (
    "/content/drive/MyDrive/fall_stress_project/"
    "processed/camera2_egofalls/video_splits_full.csv"
)
frames_root = "/content/egofalls_full_frames"
metadata_path = (
    "/content/drive/MyDrive/fall_stress_project/"
    "processed/camera2_egofalls/frame_metadata_full.csv"
)
progress_path = (
    "/content/drive/MyDrive/fall_stress_project/"
    "processed/camera2_egofalls/frame_metadata_progress.csv"
)

# Load the video split metadata
split_df = pd.read_csv(split_path)
os.makedirs(frames_root, exist_ok=True)

# Start with a clean output directory to avoid mixing old frames
if os.path.exists(frames_root):
    existing_items = os.listdir(frames_root)
    if existing_items:
        print("Existing frame folders detected.")
        print("Delete old frames manually only if they are from a failed run.")

# Resume completed videos using the progress metadata, if available
if os.path.exists(progress_path):
    previous_df = pd.read_csv(progress_path)
    completed_ids = set(previous_df["video_path"].astype(str))
    records = previous_df.to_dict("records")
    print("Previously recorded videos:", len(completed_ids))
else:
    completed_ids = set()
    records = []

saved_total = sum(int(r.get("frames_saved", 0)) for r in records)
stopped_for_space = False

for idx, row in tqdm(split_df.iterrows(), total=len(split_df)):
    video_path = str(row["video_path"])

    if video_path in completed_ids:
        continue

    # Stop if available disk space becomes too low
    free_gb = shutil.disk_usage("/content").free / (1024**3)
    if free_gb < 8:
        print(f"Stopping safely: only {free_gb:.2f} GB free.")
        stopped_for_space = True
        break

    video_name = str(row["video_name"])
    video_id = f"{idx:04d}_{os.path.splitext(video_name)[0]}"
    output_folder = os.path.join(frames_root, video_id)
    os.makedirs(output_folder, exist_ok=True)

    cap = cv2.VideoCapture(video_path)
    fps = cap.get(cv2.CAP_PROP_FPS)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    saved_count = 0

    if cap.isOpened() and fps > 0:
        interval = max(1, round(fps / 2))
        frame_index = 0

        while True:
            success, frame = cap.read()
            if not success:
                break

            if frame_index % interval == 0:
                resized = cv2.resize(frame, (224, 224))
                frame_path = os.path.join(
                    output_folder, f"frame_{saved_count:05d}.jpg"
                )

                if cv2.imwrite(
                    frame_path,
                    resized,
                    [cv2.IMWRITE_JPEG_QUALITY, 85]
                ):
                    saved_count += 1

            frame_index += 1

    cap.release()

    records.append({
        "video_name": video_name,
        "video_path": video_path,
        "dataset": row["dataset"],
        "label": row["label"],
        "activity": row["activity"],
        "camera_position": row["camera_position"],
        "split": row["split"],
        "frame_folder": output_folder,
        "frames_saved": saved_count,
        "original_fps": fps,
        "original_frames": total_frames
    })

    completed_ids.add(video_path)
    saved_total += saved_count

    if len(records) % 50 == 0:
        pd.DataFrame(records).to_csv(progress_path, index=False)
        free_gb = shutil.disk_usage("/content").free / (1024**3)
        print(
            f"\nCompleted: {len(records)}/{len(split_df)} | "
            f"Frames saved: {saved_total} | "
            f"Free space: {free_gb:.2f} GB"
        )

# Save progress even if extraction stops early
frame_df = pd.DataFrame(records)
frame_df.to_csv(progress_path, index=False)
frame_df.to_csv(metadata_path, index=False)

print("\n--- EXTRACTION SUMMARY ---")
print("Videos recorded:", len(frame_df))
print("Total frames saved:", int(frame_df["frames_saved"].sum()))
print("Videos with zero frames:", int((frame_df["frames_saved"] == 0).sum()))
print("Stopped due to low disk space:", stopped_for_space)
print("Free disk space:", f"{shutil.disk_usage('/content').free / (1024**3):.2f} GB")
print("Progress metadata:", progress_path)
print("Main metadata:", metadata_path)
```

```python
import os
import cv2
import numpy as np
import pandas as pd
from tqdm.auto import tqdm

# Define input and output paths
metadata_path = (
    "/content/drive/MyDrive/fall_stress_project/"
    "processed/camera2_egofalls/frame_metadata_full.csv"
)

output_root = (
    "/content/drive/MyDrive/fall_stress_project/"
    "processed/camera2_egofalls/sequences_full"
)

os.makedirs(output_root, exist_ok=True)

frame_df = pd.read_csv(metadata_path)

# Configure sequence generation
sequence_length = 16
image_size = 224

sequence_records = []
total_sequences = 0

# Process each video independently
for _, row in tqdm(frame_df.iterrows(), total=len(frame_df)):
    frame_folder = row["frame_folder"]

    frame_files = sorted(
        [
            f for f in os.listdir(frame_folder)
            if f.lower().endswith((".jpg", ".jpeg", ".png"))
        ],
        key=lambda name: int(
            os.path.splitext(name)[0].split("_")[-1]
        )
    )

    # Skip videos with insufficient frames
    if len(frame_files) < sequence_length:
        continue

    # Generate non-overlapping sequences within this video
    for start in range(0, len(frame_files) - sequence_length + 1,
                       sequence_length):

        selected_files = frame_files[start:start + sequence_length]
        frames = []
        valid_sequence = True

        for filename in selected_files:
            image_path = os.path.join(frame_folder, filename)
            image = cv2.imread(image_path)

            if image is None:
                valid_sequence = False
                break

            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

            if image.shape[:2] != (image_size, image_size):
                image = cv2.resize(image, (image_size, image_size))

            frames.append(image)

        if not valid_sequence or len(frames) != sequence_length:
            continue

        # Save each sequence as a separate NumPy array
        sequence_id = f"seq_{total_sequences:07d}.npy"
        sequence_path = os.path.join(output_root, sequence_id)

        np.save(sequence_path, np.stack(frames).astype(np.uint8))

        sequence_records.append({
            "sequence_id": sequence_id,
            "sequence_path": sequence_path,
            "video_name": row["video_name"],
            "video_path": row["video_path"],
            "dataset": row["dataset"],
            "label": row["label"],
            "split": row["split"],
            "activity": row["activity"],
            "camera_position": row["camera_position"],
            "start_frame_index": start,
            "sequence_length": sequence_length
        })

        total_sequences += 1

# Save sequence metadata
sequence_df = pd.DataFrame(sequence_records)

sequence_metadata_path = (
    "/content/drive/MyDrive/fall_stress_project/"
    "processed/camera2_egofalls/sequence_metadata_full.csv"
)

sequence_df.to_csv(sequence_metadata_path, index=False)

print("\n--- SEQUENCE GENERATION SUMMARY ---")
print("Total sequences:", len(sequence_df))
print("Sequences per split:")
print(sequence_df["split"].value_counts())

print("\nSequences by split and label:")
print(pd.crosstab(sequence_df["split"], sequence_df["label"]))

print("\nUnique source videos:", sequence_df["video_path"].nunique())
print("Sequence metadata saved to:", sequence_metadata_path)
print("Sequence files saved to:", output_root)
```

```python
import os
import numpy as np
import pandas as pd
import tensorflow as tf

# Set reproducible random seeds
SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

# Load sequence metadata from Google Drive
metadata_path = (
    "/content/drive/MyDrive/fall_stress_project/"
    "processed/camera2_egofalls/sequence_metadata_full.csv"
)
sequence_df = pd.read_csv(metadata_path)

# Separate the three datasets
train_df = sequence_df[sequence_df["split"] == "train"].reset_index(drop=True)
val_df = sequence_df[sequence_df["split"] == "validation"].reset_index(drop=True)
test_df = sequence_df[sequence_df["split"] == "test"].reset_index(drop=True)

# Convert class names to binary labels
label_map = {"non-fall": 0, "fall": 1}

for name, df in [
    ("train", train_df),
    ("validation", val_df),
    ("test", test_df)
]:
    if not df["label"].isin(label_map).all():
        raise ValueError(f"Unexpected labels in {name} data.")

    missing = sum(
        not os.path.isfile(path) for path in df["sequence_path"]
    )
    if missing:
        raise FileNotFoundError(
            f"{missing} sequence files are missing in {name} data."
        )

print("TensorFlow version:", tf.__version__)
print("Training sequences:", len(train_df))
print("Validation sequences:", len(val_df))
print("Test sequences:", len(test_df))

# Define a batch loader that reads sequences on demand
def load_sequence(path, label):
    def read_numpy_file(file_path):
        file_path = file_path.decode("utf-8")
        sequence = np.load(file_path).astype(np.float32) / 255.0
        return sequence

    sequence = tf.numpy_function(
        read_numpy_file,
        [path],
        tf.float32
    )
    sequence.set_shape((16, 224, 224, 3))

    label = tf.cast(label, tf.float32)
    return sequence, label

def make_dataset(df, batch_size=4, shuffle=False):
    paths = df["sequence_path"].astype(str).to_numpy()
    labels = df["label"].map(label_map).to_numpy(dtype=np.float32)

    dataset = tf.data.Dataset.from_tensor_slices((paths, labels))

    if shuffle:
        dataset = dataset.shuffle(
            buffer_size=len(df),
            seed=SEED,
            reshuffle_each_iteration=True
        )

    dataset = dataset.map(
        load_sequence,
        num_parallel_calls=2
    )

    dataset = dataset.batch(batch_size)
    dataset = dataset.prefetch(1)

    return dataset

BATCH_SIZE = 4

train_ds = make_dataset(train_df, BATCH_SIZE, shuffle=True)
val_ds = make_dataset(val_df, BATCH_SIZE)
test_ds = make_dataset(test_df, BATCH_SIZE)

# Test one batch before starting model training
for sequences, labels in train_ds.take(1):
    print("\nBatch sequence shape:", sequences.shape)
    print("Batch label shape:", labels.shape)
    print("Batch pixel range:",
          float(tf.reduce_min(sequences)),
          "to",
          float(tf.reduce_max(sequences)))
    print("Batch labels:", labels.numpy())

print("\nTraining datasets prepared successfully.")
```
